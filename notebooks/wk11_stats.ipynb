## setup

```python
# import stuff
from pathlib import Path
import pandas as pd
```

```python
# load data from folder
folder = Path('../data/processed')

# non-outlier datasets
listings_clean = pd.read_csv(folder / 'wk7_listings_clean.csv', low_memory = False)
sold_clean = pd.read_csv(folder / 'wk7_sold_clean.csv', low_memory = False)

print('Listings shape (clean):', listings_clean.shape)
print('Sold shape (clean):', sold_clean.shape)

# flagged-outlier datasets
listings_flagged = pd.read_csv(folder / 'wk7_listings_flagged.csv', low_memory = False)
sold_flagged = pd.read_csv(folder / 'wk7_sold_flagged.csv', low_memory = False)

print('Listings shape (flagged):', listings_flagged.shape)
print('Sold shape (flagged):', sold_flagged.shape)
```

```python
stats_cols = [
    'City',
    'ClosePrice',
    'DaysOnMarket',
    'price_per_sqft',
    'close_to_original_list_ratio',
    'ListOfficeName',
    'ListAgentFullName'
]
```

## check sold dataset

### clean

```python
sold_clean_mp = sold_clean[sold_clean['City'] == 'Milpitas']

print('Milpitas sold shape (clean):', sold_clean_mp.shape)
sold_clean_mp[stats_cols].head()
```

```python
sold_clean_mp[stats_cols].describe()
```

```python
sold_clean_mp['ListAgentFullName'].value_counts()
```

```python
sold_clean_mp['ListOfficeName'].value_counts()
```

```python
sold_flagged_mp = sold_flagged[sold_flagged['City'] == 'Milpitas']

print('Milpitas sold shape (flagged):', sold_flagged_mp.shape)
sold_flagged_mp[stats_cols].head()
```

```python
sold_flagged_mp[stats_cols].describe()
```
